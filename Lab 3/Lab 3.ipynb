```python
# Task 1:
""" 
Task1: Use any image processing library and apply the following geometric
transformations over the image.
1. Increase the size of the image.
2. Rotate the image by 120 degrees.
3. Perform Sheer operations over the image. 
"""

import cv2
import numpy as np

# Load image
image = cv2.imread("/Users/bosaleh_07/Desktop/Github/Image_Processing_ARTI-404_Labs/Lab 3/images/input.jpg")   # replace with your image path

# ----------------------------
# 1) Increase Size (Scaling)
# ----------------------------
scaled = cv2.resize(image, None, fx=1.5, fy=1.5)  # 1.5x bigger

# ----------------------------
# 2) Rotate 120 Degrees
# ----------------------------
(h, w) = image.shape[:2]
center = (w // 2, h // 2)

rotation_matrix = cv2.getRotationMatrix2D(center, 120, 1.0)
rotated = cv2.warpAffine(image, rotation_matrix, (w, h))

# ----------------------------
# 3) Shear Transformation
# ----------------------------
shear_factor = 0.5

shear_matrix = np.array([
    [1, shear_factor, 0],
    [0, 1, 0]
], dtype=np.float32)

sheared = cv2.warpAffine(image, shear_matrix, (int(w + shear_factor*h), h))

# ----------------------------
# Display Results
# ----------------------------
cv2.imshow("Original", image)
cv2.imshow("Scaled", scaled)
cv2.imshow("Rotated 120 deg", rotated)
cv2.imshow("Sheared", sheared)

cv2.waitKey()
```

```python
# Task 2:
"""
Task2: Use any image processing library and apply the following intensity
transformations over the image.
1. Negative
2. Log with (choose an appropriate constant for your image)
3. Power law (choose a value for gamma that increases the contrast)
"""
import cv2
import numpy as np

# Load image in grayscale (better for intensity transformations)
image = cv2.imread("/Users/bosaleh_07/Desktop/Github/Image_Processing_ARTI-404_Labs/Lab 3/images/input.jpg", 0)

# ----------------------------
# 1) Negative Transformation
# ----------------------------
negative = 255 - image


# ----------------------------
# 2) Log Transformation
# s = c * log(1 + r)
# ----------------------------
c = 255 / np.log(1 + np.max(image))
log_transformed = c * np.log(1 + image)
log_transformed = np.array(log_transformed, dtype=np.uint8)


# ----------------------------
# 3) Power Law (Gamma Correction)
# s = c * r^gamma
# gamma < 1 → brighter
# gamma > 1 → darker
# ----------------------------
gamma = 0.5   # increases contrast in darker regions
normalized = image / 255.0
power_law = np.power(normalized, gamma)
power_law = np.uint8(power_law * 255)


# ----------------------------
# Display
# ----------------------------
cv2.imshow("Original", image)
cv2.imshow("Negative", negative)
cv2.imshow("Log Transformation", log_transformed)
cv2.imshow("Power Law (Gamma=0.5)", power_law)

cv2.waitKey()
```
