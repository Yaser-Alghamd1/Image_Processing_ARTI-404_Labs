## Unsharp Masking

```python
# Read the image
image = cv2.imread('images/Parrot.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) # Convert from BGR to RGB
```

```python
# Gaussian sigma
sigma=2

#parameter
amount=1.5
```

```python
# Convert image to float32 for processing
image_float = image.astype(np.float32) / 255.0
```

```python
# Apply Gaussian blur
blurred = cv2.GaussianBlur(image_float, (0, 0), sigmaX=sigma,
sigmaY=sigma)
```

```python
# Perform Unsharp Masking
sharpened = np.clip(image_float + amount * (image_float - blurred), 0, 1)
```

```python
# Plot original, blurred, and sharpened images
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(blurred)
ax[1].set_title("Blurred Image")
ax[1].axis("off")

ax[2].imshow(sharpened)
ax[2].set_title("Sharpened (Unsharp Masking)")
ax[2].axis("off")

plt.show()
```

Task 1: 
Convolve an image with a 7x7 box filter, and output the original and smoothed
image side by side

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("images/Parrot.png")

# Convert BGR to RGB (for correct display)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Create 7x7 box filter kernel
kernel = np.ones((7, 7), np.float32) / 49

# Apply convolution
smoothed = cv2.filter2D(img_rgb, -1, kernel)

# Show original and smoothed images side by side
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(smoothed)
plt.title("Smoothed Image (7x7 Box Filter)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Task 3 :
Task#3: Using an image of your choice, apply a 3x3 Laplacian filter to sharpen the image.
Follow the following steps:
1. Read the image
2. Convert image to float32
3. Use the cv2.Laplacian() function
4. Sharpen the image using: np.clip(image_float - laplacian, 0, 1)
5. Plot the original image, the resulting image after applying the Laplacian filter, and the
sharpened image

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Read the image
img = cv2.imread("images/Parrot.png")

# Convert BGR to RGB for display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Step 2: Convert image to float32 and normalize to [0,1]
image_float = img_rgb.astype(np.float32) / 255.0

# Step 3: Apply Laplacian filter (3x3 kernel)
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

# Step 4: Sharpen image
sharpened = np.clip(image_float - laplacian, 0, 1)

# Step 5: Plot the results
plt.figure(figsize=(15,5))

# Original Image
plt.subplot(1,3,1)
plt.imshow(image_float)
plt.title("Original Image")
plt.axis("off")

# Laplacian Image
plt.subplot(1,3,2)
plt.imshow(np.clip(laplacian + 0.5, 0, 1))
plt.title("Laplacian Filter Result")
plt.axis("off")

# Sharpened Image
plt.subplot(1,3,3)
plt.imshow(sharpened)
plt.title("Sharpened Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
